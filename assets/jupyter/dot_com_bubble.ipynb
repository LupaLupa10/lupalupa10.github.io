# Getting the Fed dot plot

Below is the **exact official Federal Reserve figure** for participants' judgments of the appropriate federal funds rate path from the **December 10, 2025** Summary of Economic Projections.

This is the latest SEP available as of **March 13, 2026**.

- Federal Reserve press release: https://www.federalreserve.gov/newsevents/pressreleases/monetary20251210b.htm

How to read it:

- Each year has its own distribution of participant judgments.
- The concentration of marks shows where views cluster.
- The later-year panels show how the committee thinks policy may evolve over time.
